# Seaborn

```python
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline

plt.rcParams['figure.figsize'] = (6, 2)
plt.rcParams['font.enable_last_resort'] = False

import warnings
warnings.filterwarnings('ignore')
```

```python
sns.__version__
```

```python
df.head(1)
```

```python
vis1 = sns.distplot(df["InternetUsers"]) # plot the using 1 variable is called univeriate analysis
plt.show()
```

```python
vis2 =sns.displot(df["InternetUsers"])
plt.show()
```

```python
vis3 = sns.histplot(df["InternetUsers"])

plt.show()
```

```python
vis2 =sns.displot(df["InternetUsers"], bins = 15)
plt.show()
```

```python
vis2 =sns.displot(df["InternetUsers"], bins = 10)
plt.show()
```

```python
vis4 = sns.boxplot(data = df, x = "IncomeGroup", y = 'BirthRate')
plt.show()
```

```python
vis5 = sns.lmplot(data = df,x = 'InternetUsers', y = 'BirthRate')
plt.show()
```

```python
vis5 = sns.lmplot(data = df,x = 'InternetUsers', y = 'BirthRate' ,fit_reg = False)
plt.show()
```

```python
vis5 = sns.lmplot(data = df,x = 'InternetUsers', y = 'BirthRate' ,fit_reg = True)
plt.show()
```

```python
vis8 = sns.lmplot(data = df, x = 'InternetUsers', y= 'BirthRate', fit_reg = False, hue = 'IncomeGroup')
plt.show()
```

```python
vis8 = sns.lmplot(data = df, x = 'InternetUsers', y= 'BirthRate', fit_reg = True, hue = 'IncomeGroup')
plt.show()
```
