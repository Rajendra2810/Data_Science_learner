```python
def add_sub(x,y):
    c = x + y
    d = x - y
    return c,d 

r, r1  = add_sub(5,6) 

print(r)
print(r1)
```

```python
def add_sub_mul(x,y):
    c = x + y
    d = x - y
    e = x * y 
    return c,d,e 

r, r1,r2  = add_sub_mul(5,6) 

print(r)
print(r1)
print(r2)
```

# keyword argument
